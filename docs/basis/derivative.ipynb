# Derivatives

## Theory

The derivative of the Chebyshev polynomials can be obtained by using the following recurrence relation:
```{math}
:label: cheby_diff_rec
T'_n(x) = 2nT_{n-1}(x) + \frac{n}{n-2} T'_{n-2}(x) \;.
```
This expression is obtained by rearranging the following identity:
```{math}
2T_n(x) = \frac{T'_{n+1}(x)}{n+1} - \frac{T'_{n-1}(x)}{n-1}
```
obtained by combining equations [(18.9.9)](https://dlmf.nist.gov/18.9#E9) and [(18.9.21)](https://dlmf.nist.gov/18.9#E21) in {cite:t}`DLMFNISTDigital`.

Equation {eq}`cheby_diff_rec` can be easily generalised to higher derivatives:
```{math}
:label: cheby_diffd2
T^{(d)}_n(x) = 2nT^{(d-1)}_{n-1}(x) + \frac{n}{n-2} T^{(d)}_{n-2}(x) \;,
```
which is valid for $d>0$ and $n>2$.
This recurrence relation can be used by starting from:
```{math}
T_0(x)=1 \;, \quad T_1(x)=x \;, \quad T_2(x)=2x^2-1 \;,
```
and
```{math}
:label: cheby_diff_init
T'_0(x)=0 \;, \quad T'_1(x)=1=T_0(x) \;, \quad T'_2(x)=4x=4T_1(x) \;.
```

Consider a function $f(x)$ defined by a series of $N$ terms of Chebyshev polynomials:
```{math}
f(x) = \sum_{n=0}^{N-1} f_n T_n(x) \;.
```
where $f_n$ are the coefficients of the expansion.
We then define the function $g(x)=f'(x)$ with a similar expansion:
```{math}
g(x) = \sum_{n=0}^{N-1} g_n T_n(x) \;.
```
The coefficients $f_n$ and $g_n$ can be stored in the vectors $\vec{f}$ and $\vec{g}$ of length $N$.
They are related  by the following equation:
```{math}
\vec{g} = \mat{D} \vec{f} \;,
```
where $\mat{D}$ is the differentiation matrix.
From equation {eq}`cheby_diff_init`, we can see that the first three columns of $\mat{D}$ are:
```{math}
\begin{pmatrix}
0 & 1 & 0 \\
0 & 0 & 4 \\
0 & 0 & 0 \\
\vdots & \vdots & \vdots
\end{pmatrix} \;.
```
The subsequent columns are calculated using the recurrence relation {eq}`cheby_diff_rec`.

## Python API

The `Basis1D` class provides two methods related to the derivatives of the Chebyshev polynomials.

Firstly, the method `derivatives` computes the derivatives of the Chebyshev polynomials up to a given order at a given set of points $x_n$.
For instance, to plot the first and second derivatives of the first 10 Chebyshev polynomials, one can use:

````python
from cheby import Basis1D
import numpy as np
import matplotlib.pyplot as plt
%config InlineBackend.figure_formats = ["svg", "pdf"]

x = np.linspace(-1, 1, 300)

basis = Basis1D(10)
dTn = basis.derivatives(x, 2)

plt.figure(figsize=(9, 3))

plt.subplot(1,2,1)
plt.plot(x, dTn[1])
plt.xlabel(r'$x$')
plt.title('first derivative')

plt.subplot(1,2,2)
plt.plot(x, dTn[2])
plt.xlabel(r'$x$')
plt.title('second derivative')

plt.tight_layout()
plt.show()
````

Secondly, the method `diff_matrix` returns the differentiation matrix $\mat{D}$.
For instance, to calculate the differentiation matrix for a Chebyshev series of order 5 (hence with 6 terms) defined on the interval $[0,1]$, one can use:

````python
from cheby import Basis1D

basis = Basis1D(5, 0, 1)
D = basis.diff_matrix()
print(D)
````
