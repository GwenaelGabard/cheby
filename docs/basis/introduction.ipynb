# Basis

This section describes the basis of Chebyshev polynomials. 
For a detailed description of the Chebyshev polynomials please see {cite:t}`masonChebyshevPolynomials2003` and {cite:t}`trefethenApproximationTheoryApproximation2020`.


## Theory

The Chebyshev polynomials are defined on the interval $[-1;+1]$ by the following expression:
```{math}
T_n(x) = \cos[n \arccos(x)]
```
where the integer $n$ is the order of the polynomial and $x$ is the independent variable.

This family of polynomials can also be defined through the following recursive relation:
```{math}
:label: rec_cheby
T_{n+1}(x) = 2xT_n(x) - T_{n-1}(x) \;,
```
for $n \geq 1$, together with $T_0(x)=1$ and $T_1(x)=x$.

It is easy to map the interval $[-1;+1]$ to any other interval $[a;b]$ using the following transformation:
```{math}
:label: xi_map
x = \frac{a+b}{2} + \frac{b-a}{2} \xi \;,
```
where $\xi$ is the independent variable in the interval $[-1;+1]$.


## Python API

A basis of Chebyshev polynomials of fixed order can be created using the class `Basis1D` from `cheby`.
One has to specify the order $N$ of the basis and the bounds of the domain $[a;b]$.
For instance

````python
from cheby import Basis1D

basis = Basis1D(7, 0, 3)
````

creates a basis of Chebyshev polynomials of maximal order 7 (hence with $N=8$ basis functions) defined on the interval $[0, 3]$ (hence with $a=0$ and $b=3$).

It is possible to access the order of the basis through the attribute `order`, as well as the bounds of the interval through the attributes `start` and `end`:

````python
order = basis.order
x_min = basis.start
x_max = basis.end
````



It is easy to evaluate the $N$ Chebyshev polynomials at a set of points $x_n$ on the interval $[a;b]$ by calling the `eval` method of the basis object:

````python
import numpy as np
import matplotlib.pyplot as plt
%config InlineBackend.figure_formats = ["svg", "pdf"]

x = np.linspace(0, 3, 300)

Tn = basis.eval(x)

plt.figure()
plt.plot(x, Tn)
plt.xlabel(r'$x$')
plt.show()
````

The implementation of the `eval` method first applies the inverse of the mappring {eq}`xi_map`:
```{math}
\xi_n = 2\frac{x_n - a}{b-a} - 1 \;,
```
where $x_n$ are the parameter values provided as input.
Then the recursive relation {eq}`rec_cheby` is used to compute the values of the Chebyshev polynomials.
