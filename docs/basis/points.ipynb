# Points

This section describes the basis of Chebyshev polynomials. 
For a detailed description of the Chebyshev polynomials please see {cite:t}`masonChebyshevPolynomials2003` and {cite:t}`trefethenApproximationTheoryApproximation2020`.

## Theory

Two families of collocation points are commonly used with Chebyshev polynomials.

The **points of the first kind**, also called the Gauss–Chebyshev points, are the roots of $T_N(\xi)$:
```{math}
:label: points1_std
\xi_k = \cos\!\left(\frac{(2k+1)\pi}{2N}\right) \;, \qquad k = 0, \dots, N-1 \;.
```

The **points of the second kind**, also called the Chebyshev–Lobatto points, are the extrema of $T_{N-1}(\xi)$:
```{math}
:label: points2_std
\xi_k = \cos\!\left(\frac{k\pi}{N-1}\right) \;, \qquad k = 0, \dots, N-1 \;.
```

Unlike the points of the first kind, the points of the second kind do include the two endpoints $\pm 1$.
Both families are mapped from $[-1;+1]$ to the interval $[a;b]$ of the basis using the mapping {eq}`xi_map`.
In both cases, the points cluster near the ends of the interval.


## Python API

The `Basis1D` class provides the methods `points1` and `points2` to generate the two families of points on the interval of the basis. By default the number of points equals `order + 1`, but a different number of points can be requested explicitly.

````python
import numpy as np
from cheby import Basis1D

basis = Basis1D(15, -1, 1)
x1 = basis.points1()
x2 = basis.points2()

print(f'{len(x1)} points of the first kind')
print(f'{len(x2)} points of the second kind (includes the endpoints)')
````

The clustering near the endpoints can be visualised by projecting equally spaced angles on the unit semicircle onto the $x$-axis, which is exactly the construction behind {eq}`points1_std` and {eq}`points2_std`.

````python
import matplotlib.pyplot as plt
%config InlineBackend.figure_formats = ["svg", "pdf"]

theta = np.linspace(0, np.pi, 200)

fig, axes = plt.subplots(1, 2, figsize=(9, 4.5))
for ax, x, title in zip(axes, [x1, x2], ['first kind', 'second kind']):
    ax.plot(np.cos(theta), np.sin(theta), 'lightgray')
    angles = np.arccos(x)
    ax.plot(x, np.sin(angles), 'o', color='C0')
    ax.vlines(x, 0, np.sin(angles), color='C0', linewidth=0.5)
    ax.plot(x, np.zeros_like(x), 'x', color='C1')
    ax.set_title(title)
    ax.set_aspect('equal')
    ax.set_xlabel(r'$\xi$')
plt.tight_layout()
plt.show()
````

The crosses on the $\xi$-axis are the Chebyshev points themselves: they are denser near the endpoints $\xi=\pm1$ compared to the centre of the interval.
