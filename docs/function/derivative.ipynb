# Derivative

This section describes how to compute the derivative of a Chebyshev series.

## Theory

If we consider a function defined as a Chebyshev series:
```{math}
f(x) = \sum_{n=0}^{N} c_n T_n(x)
\;,
```
its derivative is directly given as
```{math}
f'(x) = \sum_{n=0}^{N} c_n T'_n(x) \;.
```
This is not in the form of a Chebyshev series because the derivative of a Chebyshev polynomial is not a Chebyshev polynomial.
However, it can be expressed as a linear combination of Chebyshev polynomials.
For this, we write:
```{math}
f'(x) = \sum_{n=0}^{N} d_n T_n(x) \;,
```
where the coefficients $d_n$ are related to the coefficients $c_n$ by a linear transformation:
```{math}
:label: func_derivative
\vec{d} = \mat{D}\,\vec{c} \;,
```
where $\vec{c}$ and $\vec{d}$ are the coefficient vectors of $f$ and $f'$, padded with zeros to a common length.
The differentiation matrix $\mat{D}$ can be obtained from the method `diff_matrix` of the class `Basis1D`, which is based on the recurrence relation {eq}`cheby_diff_rec`.
However, it is not necessary to build $\mat{D}$ explicitly because the recurrence can be applied directly to the coefficients $\vec{c}$.
This is done by using equations (2.50) and (2.51) in {cite:t}`masonChebyshevPolynomials2003`.

## Python API

As an example we construct a Chebyshev series for the function $f(x) = \sin(3x)$ on the interval $[0, 2]$:

````python
import numpy as np
from cheby import RealFunction

f = RealFunction(lambda x: np.sin(3 * x), 0.0, 2.0)
````

The method `derivative` of the class `RealFunction` returns the derivative function $f'$, which is also an instance of `RealFunction`:

````python
df = f.derivative()
type(df)
````

We can then compare with the exact derivative $f'(x) = 3\cos(3x)$:

````python
import matplotlib.pyplot as plt
%config InlineBackend.figure_formats = ["svg", "pdf"]

x = np.linspace(0.0, 2.0, 300)
plt.plot(x, df(x), label="f.derivative()")
plt.plot(x, 3 * np.cos(3 * x), '--', label="exact f'")
plt.xlabel("x")
plt.legend()
plt.show()
````

The coefficients of the derivative could have been calculated by using the differentiation matrix $\mat{D}$ provided by the method `diff_matrix` of the class `Basis1D`.
We can check easily that the result is identical to the one obtained by using the `derivative` method:

````python
df_coef = f.basis().diff_matrix() @ f.coef

np.allclose(df_coef[:-1], df.coef)
````
