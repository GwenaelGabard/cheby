# Functions as Chebyshev series

This chapter introduces the `RealFunction` and `ComplexFunction` classes, which represent univariate functions over a closed interval $[a, b]$ as truncated Chebyshev series:
```{math}
:label: cheb_series
f(x) = \sum_{n=0}^{N} c_n T_n(x), \quad x \in [a, b]
\;,
```
where $T_n$ are the Chebyshev polynomials of the first kind and $c_n$ are the Chebyshev coefficients.
The function $f$ is therefore a polynomial of order $N$.

All of the operations covered in the rest of this chapter (evaluation, arithmetic, calculus, roots, norms, ...) are implemented purely in terms of the coefficients $c_n$, without resampling the function, which is what makes spectral methods based on Chebyshev series efficient: differentiation, integration and multiplication all reduce to (comparatively cheap) linear or bilinear operations on short coefficient vectors.

The theory and applications of Chebyshev polynomials are described in detail in the following textbooks:
* {cite:t}`masonChebyshevPolynomials2003`
* {cite:t}`boyd01`
* {cite:t}`trefethenApproximationTheoryApproximation2020`
* {cite:t}`rivlinChebyshevPolynomialsApproximation2020`

The classes `RealFunction` or `ComplexFunction` represent functions of the form {eq}`cheb_series` by storing the interval bounds $a$ and $b$ in the variables `start` and `end`, and the coefficient $c_n$ in a vector `coef`.
`RealFunction` and `ComplexFunction` share the same interface; the only difference is the scalar type of the coefficients (`float64` vs `complex128`). Combining a `RealFunction` and a `ComplexFunction` in an arithmetic operation (see {doc}`arithmetic`) automatically promotes the result to a `ComplexFunction`.
The following sections describe how to construct these representations and what operations are available.

## References

```{bibliography}
:filter: docname in docnames
```
