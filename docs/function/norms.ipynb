# Norms

This section describes the $L^2$ and $H^1$ norms of a Chebyshev series.

## Theory

The $L^2([a,b])$ norm of $f$ is defined as usual by
```{math}
\|f\|_{L^2}^2 = \int_a^b |f(x)|^2\,\mathrm{d}x = \int_a^b f(x)\,\overline{f(x)}\,\mathrm{d}x \;.
```
This can be evaluated exactly for a Chebyshev series.
First, by computing the product $f \bar f$, as explained in {doc}`arithmetic`.
Then the integral is computed using the closed-form expression {eq}`func_integral` from {doc}`primitives`.

The $H^1([a,b])$ norm is essentially a combination of the $L^2$ norm of the function and its derivative:
```{math}
\|f\|_{H^1}^2 = \|f\|_{L^2}^2 + \alpha\,\|f'\|_{L^2}^2
\;,
```
with a weight $\alpha$ (defaulting to $1$) controlling how strongly the derivative contributes.
The derivative $f'$ is computed with {doc}`derivative`, and its $L^2$ norm is computed as above.

## Python API

````python
import numpy as np
from cheby import RealFunction

f = RealFunction(lambda x: np.sin(np.pi * x), 0.0, 1.0)

print('||f||_L2  =', f.norm_L2(), ' (exact:', 1 / np.sqrt(2), ')')
print('||f||_H1  =', f.norm_H1())
print('||f||_H1 (alpha=0) == ||f||_L2 :',
      np.isclose(f.norm_H1(alpha=0.0), f.norm_L2()))
````

For this example the exact $H^1$ norm can also be worked out analytically, since $f(x)=\sin(\pi x)$ and $f'(x)=\pi\cos(\pi x)$ both have $L^2([0,1])$ norm $1/\sqrt{2}$ and $\pi/\sqrt{2}$ respectively:

````python
alpha = 1.0
exact_h1 = np.sqrt(0.5 + alpha * (np.pi**2) * 0.5)
print('computed:', f.norm_H1(alpha=alpha))
print('exact:   ', exact_h1)
````

`ComplexFunction` also provides `norm_L2`/`norm_H1`; the complex conjugate in the definition above ensures both norms are always non-negative real numbers, even for complex-valued functions.

````python
from cheby import ComplexFunction

cf = ComplexFunction(lambda x: np.exp(1j * np.pi * x), 0.0, 1.0)
print(cf.norm_L2(), '(exact: 1.0)')
````
