# Arithmetic operations

This section describes addition, subtraction, multiplication, and integer powers of Chebyshev series.

## Theory

Addition and subtraction of two Chebyshev series are simply coefficient-wise addition and subtraction, padding the shorter coefficient vector with zeros where needed.
There is no coupling between different Chebyshev polynomials in a sum.
Both functions must share the same interval $[a, b]$.

Multiplication is less trivial, because the product of two Chebyshev polynomials is not a single Chebyshev polynomial.
We can use the product formula, described in Section 2.4.3 in {cite:t}`masonChebyshevPolynomials2003`:
```{math}
:label: cheby_product
T_m(\xi) \, T_n(\xi) = \tfrac{1}{2}\bigl(T_{m+n}(\xi) + T_{|m-n|}(\xi)\bigr) \;.
```
The product of two series
```{math}
f(x) = \sum_{m=0}^{M} f_m T_m(x) \;, \qquad
g(x) = \sum_{n=0}^{N} g_n T_n(x)
```
expands to
```{math}
:label: cheby_multiply
f(x)\,g(x) = \frac{1}{2}\sum_{m=0}^{M}\sum_{n=0}^{N} f_m g_n \Bigl(T_{m+n}(x) + T_{|m-n|}(x)\Bigr) \;.
```
The coefficients of the product are obtained by distributing each pairwise product $f_m g_n$ onto the two entries $m+n$ and $|m-n|$ of the result.
The resulting series has up to $M+N+1$ coefficients, and is trimmed as described in {doc}`construction`.

Integer powers $f^n$ are computed efficiently by repeated squaring using {eq}`cheby_multiply`.

## Python API

We start by defining two `RealFunction` objects:

````python
import numpy as np
from cheby import RealFunction

f = RealFunction(lambda x: np.sin(3 * x), 0.0, 2.0)
g = RealFunction(lambda x: np.exp(-0.5 * x), 0.0, 2.0)
````

We can then add, subtract, and multiply them:

````python
s = f + g
d = f - g
p = f * g
````

Both `RealFunction` and `ComplexFunction` support the standard arithmetic operators.
Mixing a real and a complex function promotes the result to `ComplexFunction`:

````python
from cheby import ComplexFunction

cf = ComplexFunction(lambda x: 1j * np.exp(-0.5 * x), 0.0, 2.0)
sc = f + cf
type(sc)
````

Integer powers use `**` or the equivalent `pow` method:

````python
cube = f**3
cube2 = f.pow(3)
print(np.allclose(cube.coef, cube2.coef))
````

All these operations can be combined:

````python
h = (f + g) * f**3
h.coef
````
