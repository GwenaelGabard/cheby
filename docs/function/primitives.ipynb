# Primitives and integrals

This section describes how to compute the integral of a Chebyshev series, both over the whole domain and over a sub-interval.

## Theory

Consider a Chebyshev series of the form
```{math}
f(x) = \sum_{n=0}^{N} c_n T_n(x)
\;.
```

We can write the primitive (or antiderivative) of $f$ as:
```{math}
F(x) = \int f(x)dx = \frac{b-a}{2} \int f(\xi)d\xi = \sum_{n=0}^{N} g_n T_n(x)
\;.
```
The coefficients $g_n$ are calculated by using equation (2.43) in {cite:t}`masonChebyshevPolynomials2003`.

The integral of $f$ over an arbitrary sub-interval $[\alpha, \beta] \subseteq [a, b]$ is instead obtained from the fundamental theorem of calculus:
```{math}
\int_{\alpha}^{\beta} f(x)\,\mathrm{d}x = F(\beta) - F(\alpha) \;,
```
where $F$ is a primitive (or antiderivative) of $f$.

For the special case of integral over the whole interval $[a, b]$, the integral can be integrated term by term using {eq}`cheby_integral_ab`.
This gives a closed form:
```{math}
:label: func_integral
\int_a^b f(x)\,\mathrm{d}x = (b-a)\left[c_0 + \sum_{\substack{n=2 \\ n \text{ even}}}^{N} \frac{c_n}{1-n^2}\right]
\;.
```
The odd-order terms integrate to zero since they are anti-symmetric over the interval.

## Python API

Let us use the function $f(x) = 1+\sin(3 x)$ on the interval $[0, 2]$ as an example.
We first construct the Chebyshev series interpolating $f$.

````python
import numpy as np
from cheby import RealFunction

def f_exact(x):
    return 1.0 + np.sin(3 * x)

f = RealFunction(f_exact, 0.0, 2.0)
````

The primitive of $f$ is obtained by using the method `primitive` of the class `RealFunction`:

````python
F = f.primitive()
type(F)
````

The primitive of a function is not unique because it is defined up to a constant.
The implementation of the method `primitive` simply leave the constant term $g_0$ equal to zero.

````python
F.coef
````

We can check that the derivative of the primitive is the original function:

````python
np.allclose(F.derivative().coef, f.coef)
````

The integral over the whole interval is computed with the `integral()` method:

````python
I = f.integral()
print(f"Integral: {I:.10f}")
````

We can compare against the numerical integration from `scipy`:

````python
from scipy import integrate

I_ref = integrate.quad(f_exact, 0.0, 2.0)[0]
print(f'Reference value: {I_ref:.10f}')
print(f'Difference: {I - I_ref:.5e}')
````

We can compute the integral over a sub-interval $[\alpha, \beta]$ using the same `integral()` method with two arguments:

````python
sub = f.integral(0.5, 1.5)
sub_ref = integrate.quad(f_exact, 0.5, 1.5)[0]
print(f'Sub-interval integral: {sub:.10f}')
print(f'Sub-interval reference: {sub_ref:.10f}')
print(f'Sub-interval difference: {sub - sub_ref:.5e}')
````

This implementation relies on the `primitive()` method to compute the antiderivative of the Chebyshev series, and then evaluates it at the endpoints of the sub-interval.

`ComplexFunction` supports the same two forms of `integral`, returning a complex value.
